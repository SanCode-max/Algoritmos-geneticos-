```python
import numpy as np 

#Ejercicio 1: Portafolio de Inversiones

# Datos del problema (10 proyectos)
costos = np.array([10, 15, 20, 25, 30, 35, 40, 45, 50, 55])
retornos = np.array([15, 25, 30, 40, 45, 55, 60, 70, 75, 85])
presupuesto_max = 100

def decodificar(cromosoma):
    return np.array(cromosoma)

def calcular_aptitud_ejercicio1(cromosoma, penalización_fuerte=True):
    genes = decodificar(cromosoma)
    costo_total = np.sum(genes * costos)
    retorno_total = np.sum(genes * retornos)
    
    if costo_total > presupuesto_max:
        exceso = costo_total - presupuesto_max
        if penalización_fuerte:
            # Penalización fuerte: invalida severamente el fitness
            return max(0, retorno_total - (exceso * 50))
        else:
            # Penalización suave: resta proporcionalmente
            return max(0, retorno_total - (exceso * 2))
    
    return retorno_total

# Ejemplo de uso
ejemplo_ind_e1 = [1, 0, 1, 1, 0, 1, 0, 0, 0, 0] # Costo: 10+20+25+35 = 90 (Válido)
print("Aptitud Ejercicio 1 (Válido):", calcular_aptitud_ejercicio1(ejemplo_ind_e1))
```
